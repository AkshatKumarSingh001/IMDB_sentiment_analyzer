```python
## Step 1: Import Librabries and load the model
import numpy as np
import keras
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
## Load the IMDB dataset word index
word_index = imdb.get_word_index()
reverse_word_index = {value: key for (key, value) in word_index.items()}
```

```python
## Load the saved model
model = load_model('simple_rnn_model.h5')
model.summary()
```

```python
model.get_weights()
```

```python
## Step 2: Helper function to decode the review
def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in encoded_review])

# Function to process user input
def process_text(text):
    words = text.lower().split()
    encoded_review = [word_index.get(word,2) + 3 for word in words]  # +3 to account for reserved indices
    padded_review = sequence.pad_sequences([encoded_review], maxlen=500)
    return padded_review
```

```python
### Prediction function

def predict_review(review):
    processed_review = process_text(review)
    prediction = model.predict(processed_review)
    sentiment = "Positive" if prediction[0][0] > 0.5 else "Negative"
    return sentiment, prediction[0][0]
```

```python
### Step 3: User input for prediction

example_review = "The movie was fantastic! The acting was great and the plot was thrilling but a bit predictable. I would definitely recommend it to my friends."

sentiment, score = predict_review(example_review)

print(f"Review: {example_review}")
print(f"Predicted Sentiment: {sentiment} (Score: {score:.4f})")
```
